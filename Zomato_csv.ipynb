```python

# ============================================================
# ZOMATO: PREDICT ONLINE DELIVERY AVAILABILITY
# Optimized, leakage-aware ML pipeline
# ============================================================

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score, classification_report,
    ConfusionMatrixDisplay, RocCurveDisplay
)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier

# -------------------- 1. CONFIGURATION --------------------

TARGET = "Has Online delivery"
RANDOM_STATE = 42
TEST_SIZE = 0.20
CV_FOLDS = 5

if TARGET not in zomato.columns:
    raise KeyError(f"Target column {TARGET!r} not found.")

# -------------------- 2. CLEAN DATA --------------------

data = zomato.copy()

# Normalize target safely
data[TARGET] = (
    data[TARGET]
    .astype("string")
    .str.strip()
    .str.lower()
)

data[TARGET] = data[TARGET].replace({
    "yes": "Yes",
    "no": "No"
})

data = data[data[TARGET].isin(["Yes", "No"])].copy()

if data.empty or data[TARGET].nunique() != 2:
    raise ValueError("Target must contain both Yes and No.")

# Exclude target, identifiers, post-outcome labels and operational proxy
exclude = [
    TARGET,
    "Restaurant ID",
    "Restaurant Name",
    "Address",
    "Locality Verbose",
    "Rating text",
    "Rating color",
    "Is delivering now"
]

X = data.drop(
    columns=[c for c in exclude if c in data.columns]
).copy()

y = data[TARGET].eq("Yes").astype("int8")

# Convert pandas nullable strings to ordinary Python objects,
# replacing missing/blank values with np.nan.
# This avoids pd.NA ambiguity in downstream estimators.
for col in X.select_dtypes(include=["object", "string"]).columns:
    s = X[col].astype(object)

    def clean_value(v):
        if pd.isna(v):
            return np.nan
        if isinstance(v, str):
            v = v.strip()
            if v.lower() in {"", "nan", "none", "<na>"}:
                return np.nan
            return v
        return v

    X[col] = s.map(clean_value)

# Remove columns that contain no usable values at all
empty_cols = X.columns[X.isna().all()].tolist()
X = X.drop(columns=empty_cols)

if X.shape[1] == 0:
    raise ValueError("No usable features remain after cleaning.")

# -------------------- 3. TRAIN / TEST SPLIT --------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

# Detect feature types using training data only
numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = [
    c for c in X_train.columns if c not in numeric_features
]

# -------------------- 4. PREPROCESSING --------------------

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        min_frequency=10,
        sparse_output=True
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# -------------------- 5. MODELS --------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1500,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=150,
        min_samples_leaf=2,
        class_weight="balanced_subsample",
        n_jobs=1,
        random_state=RANDOM_STATE
    ),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=150,
        min_samples_leaf=2,
        class_weight="balanced",
        n_jobs=1,
        random_state=RANDOM_STATE
    )
}

# -------------------- 6. CROSS-VALIDATION --------------------

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_rows = []
fitted_models = {}

for name, estimator in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", estimator)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        error_score="raise",
        return_train_score=False
    )

    row = {"Model": name}

    for metric in scoring:
        values = scores[f"test_{metric}"]
        row[f"CV {metric}"] = values.mean()
        row[f"CV {metric} SD"] = values.std()

    cv_rows.append(row)

    # Fit one final pipeline on all training data
    pipeline.fit(X_train, y_train)
    fitted_models[name] = pipeline

cv_results = (
    pd.DataFrame(cv_rows)
    .sort_values("CV roc_auc", ascending=False)
    .reset_index(drop=True)
)

display(
    cv_results.style.format({
        c: "{:.3f}" for c in cv_results.columns
        if c != "Model"
    })
)

# -------------------- 7. HOLDOUT EVALUATION --------------------

best_name = cv_results.loc[0, "Model"]
best_model = fitted_models[best_name]

y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

holdout_metrics = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Balanced Accuracy",
        "Precision (Yes)",
        "Recall (Yes)",
        "F1 (Yes)",
        "ROC-AUC"
    ],
    "Score": [
        accuracy_score(y_test, y_pred),
        balanced_accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred, zero_division=0),
        recall_score(y_test, y_pred, zero_division=0),
        f1_score(y_test, y_pred, zero_division=0),
        roc_auc_score(y_test, y_prob)
    ]
})

print(f"Selected model: {best_name}")
display(holdout_metrics.style.format({"Score": "{:.4f}"}))

print("\nClassification Report:\n")
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1],
    target_names=["No online delivery", "Online delivery"],
    zero_division=0
))

# -------------------- 8. VISUALIZATION --------------------

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    labels=[0, 1],
    display_labels=["No", "Yes"],
    cmap="Blues",
    ax=axes[0],
    colorbar=False
)
axes[0].set_title(f"Confusion Matrix — {best_name}")

RocCurveDisplay.from_predictions(
    y_test,
    y_prob,
    ax=axes[1],
    name=best_name
)
axes[1].plot(
    [0, 1], [0, 1],
    linestyle="--",
    linewidth=1,
    label="Random baseline"
)
axes[1].set_title("Holdout ROC Curve")
axes[1].legend()

plt.tight_layout()
plt.show()

# -------------------- 9. SUMMARY --------------------

print(f"Original rows: {len(zomato):,}")
print(f"Rows used: {len(data):,}")
print(f"Training rows: {len(X_train):,}")
print(f"Holdout rows: {len(X_test):,}")
print(f"Features used: {X.shape[1]}")
print("Removed all-missing features:", empty_cols)
print("Excluded columns:", [c for c in exclude if c in zomato.columns])
```
